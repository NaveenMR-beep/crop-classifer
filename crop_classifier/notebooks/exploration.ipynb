# 🌾 Crop Image Dataset — Exploratory Analysis

This notebook explores the CICR crop image dataset before model training.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

DATA_DIR = Path('../data')
CSV_PATH = DATA_DIR / 'Crop_details.csv'

df = pd.read_csv(CSV_PATH, index_col=0)
print(df.shape)
df.head()
```

```python
# Class distribution
counts = df['crop'].value_counts()
counts.plot.bar(color='green', figsize=(8, 4))
plt.title('Samples per Crop Class')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

```python
# Sample images
from PIL import Image
import random

CROPS = ['jute', 'maize', 'rice', 'sugarcane', 'wheat']
fig, axes = plt.subplots(1, 5, figsize=(15, 3))

for ax, crop in zip(axes, CROPS):
    img_dir = DATA_DIR / 'crop_images' / crop
    imgs = list(img_dir.glob('*.jpeg'))
    if imgs:
        img = Image.open(random.choice(imgs))
        ax.imshow(img)
    ax.set_title(crop.title())
    ax.axis('off')

plt.suptitle('One Sample per Class', fontsize=14)
plt.tight_layout()
plt.show()
```

```python
# Pixel statistics
import tensorflow as tf

means, stds = [], []
for crop in CROPS:
    img_dir = DATA_DIR / 'crop_images' / crop
    for path in list(img_dir.glob('*.jpeg'))[:10]:
        img = tf.image.decode_jpeg(tf.io.read_file(str(path)), channels=3)
        img = tf.cast(img, tf.float32) / 255.0
        means.append(tf.reduce_mean(img, axis=[0, 1]).numpy())
        stds.append(tf.math.reduce_std(img, axis=[0, 1]).numpy())

print('Dataset mean (R,G,B):', np.mean(means, axis=0).round(3))
print('Dataset std  (R,G,B):', np.mean(stds,  axis=0).round(3))
```
